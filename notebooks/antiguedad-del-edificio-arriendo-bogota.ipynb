Este cuaderno carga los datos abiertos del estudio y reproduce su tabla y gráfica principal.

```python
import os, pandas as pd, matplotlib.pyplot as plt
RUTA_KAGGLE = '/kaggle/input/bogota-2026-antiguedad-del-edificio-arriendo/antiguedad_del_edificio_arriendo_bogota_datos.csv'
RUTA_GITHUB = 'https://raw.githubusercontent.com/elbroker/aptologia-datos/main/estudios/antiguedad-del-edificio-arriendo-bogota/data/antiguedad_del_edificio_arriendo_bogota_datos.csv'
df = pd.read_csv(RUTA_KAGGLE if os.path.exists(RUTA_KAGGLE) else RUTA_GITHUB)
print(df.shape)
df.head(10)
```

## Tabla principal

```python
t = df
if 'tabla' in df.columns:
    print('Tablas disponibles:', df['tabla'].unique().tolist())
    otras = [x for x in df['tabla'].unique() if x != 'bogota']
    t = df[df['tabla'] == otras[0]] if otras else df
t = t.dropna(axis=1, how='all')
t
```

## Gráfica principal

```python
pref = ['mediana','indice_precio_calidad','canon_mediano','prima_m2','m2_mediano','total_mensual_mediano','porcentaje','m2_por_millon']
num = [c for c in pref if c in t.columns] or t.select_dtypes('number').columns.tolist()
y = num[0]
cand = [c for c in t.columns if not pd.api.types.is_numeric_dtype(t[c]) and c not in ('tabla','barrio_slug','pagina')]
x = cand[0] if cand else t.columns[0]
g = t[[x, y]].dropna().copy()
g[x] = g[x].astype(str)
if len(cand) > 1 and cand[1] != x and t[cand[1]].nunique() < 6:
    g[x] = g[x] + ' · ' + t.loc[g.index, cand[1]].astype(str)
if len(g) > 20:
    g = g.sort_values(y, ascending=False).head(20)
ax = g.plot.barh(x=x, y=y, figsize=(9, max(3, 0.4*len(g))), legend=False, color='#1f6f5c')
ax.invert_yaxis(); ax.set_xlabel(y); ax.set_ylabel('')
ax.set_title('Edificio nuevo o viejo: cuánto cambia el arriendo según la antigüedad en Bogotá (2026)', fontsize=10)
plt.tight_layout(); plt.show()
```
