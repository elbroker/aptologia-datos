Este cuaderno carga los datos abiertos del estudio y reproduce su tabla y gráfica principal.

```python
import os, pandas as pd, matplotlib.pyplot as plt
RUTA_KAGGLE = '/kaggle/input/arriendo-en-rionegro-por-barrio-2026/arriendo_en_rionegro_por_barrio_2026_datos.csv'
RUTA_GITHUB = 'https://raw.githubusercontent.com/elbroker/aptologia-datos/main/estudios/arriendo-en-rionegro-por-barrio-2026/data/arriendo_en_rionegro_por_barrio_2026_datos.csv'
df = pd.read_csv(RUTA_KAGGLE if os.path.exists(RUTA_KAGGLE) else RUTA_GITHUB)
print(df.shape)
df.head(10)
```

## Tabla principal

```python
t = df
if 'tabla' in df.columns:
    print('Tablas disponibles:', df['tabla'].unique().tolist())
    t = df[df['tabla'] == df['tabla'].iloc[0]]
t = t.dropna(axis=1, how='all')
t
```

## Gráfica principal

```python
pref = ['rent_bruta_todos_pct_calc','mediana','salarios_minimos','prima_m2','dif_m2_vs_otros','dif_m2_vs_inmobiliaria','dif_canon','dif_m2_casa_vs_apto','canon_mediano','m2_mediano','porcentaje']
num = [c for c in pref if c in t.columns] or t.select_dtypes('number').columns.tolist()
y = num[0]
cand = [c for c in t.columns if not pd.api.types.is_numeric_dtype(t[c]) and c not in ('tabla','ciudad','fecha_datos','barrio_id')]
x = cand[0] if cand else t.columns[0]
g = t[[x, y]].dropna().copy()
g[x] = g[x].astype(str)
if len(cand) > 1 and cand[1] != x and t[cand[1]].nunique() < 6:
    g[x] = g[x] + ' · ' + t.loc[g.index, cand[1]].astype(str)
if len(g) > 20:
    g = g.sort_values(y, ascending=False).head(20)
ax = g.plot.barh(x=x, y=y, figsize=(9, max(3, 0.4*len(g))), legend=False, color='#1f6f5c')
ax.invert_yaxis(); ax.set_xlabel(y); ax.set_ylabel('')
ax.set_title('Arriendo en Rionegro por barrio: precios por alcobas, barrios más baratos y más caros (octubre 2026)', fontsize=10)
plt.tight_layout(); plt.show()
```
